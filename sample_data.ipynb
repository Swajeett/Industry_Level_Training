```python
import pandas as pd
df = pd.read_csv("sample_data.csv")   # read CSV

print (df)

print (df.head())
print (df.tail())
print (df.info())
print (df.describe())
print (df.shape)
print (df.columns)
```

```python
# calculate and display the mean price
mean_price = df['Price'].mean()
print(f"Mean Price: {mean_price}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(df['Product'], df['Price'], marker='o')
plt.axhline(y=mean_price, color='r', linestyle='-', label=f'Mean Price: {mean_price:.2f}')
plt.xlabel('Product')
plt.ylabel('Price')
plt.title('Price of Products with Mean Price')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure (figsize=(6,10))
plt.plot(df['Price'], df['Product'], marker='o')
plt.axvline(x=mean_price, color='black', linestyle='-', label=f'Mean Price: {mean_price:.2f}')
plt.xlabel('Price')
plt.ylabel('Product')
plt.title('Price of Products with Mean Price')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```
