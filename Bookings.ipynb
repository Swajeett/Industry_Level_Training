```python
df['Canceled_Rides_by_Customer'] = df['Canceled_Rides_by_Customer'].fillna('NA')
df['Canceled_Rides_by_Customer'].isnull().sum()
print(df.isnull().sum())
```

```python
df['Canceled_Rides_by_Driver'] = df['Canceled_Rides_by_Driver'].fillna('NA')
df['Canceled_Rides_by_Driver'].isnull().sum()
print(df.isnull().sum())
```

```python
df['Incomplete_Rides'] = df['Incomplete_Rides'].fillna('NA')
df['Incomplete_Rides'].isnull().sum()
print(df.isnull().sum())
```

```python
df['Incomplete_Rides_Reason'] = df['Incomplete_Rides_Reason'].fillna('NA')
df['Incomplete_Rides_Reason'].isnull().sum()
print(df.isnull().sum())
```

```python
df['Payment_Method'] = df['Payment_Method'].fillna('NA')
df['Payment_Method'].isnull().sum()
print(df.isnull().sum())
```

```python
df_filled = df.fillna(df.mean(numeric_only=True))
print("\nMissing Values After")
print(df_filled.isnull().sum())
df_filled.to_csv('Cleaned_Bookings.csv', index=False)
```
