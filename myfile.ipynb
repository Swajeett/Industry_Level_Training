```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re

df = pd.read_csv('my_file.csv')
df
```

```python
df.head()
df.tail()
df.info()
df.shape
df.describe()
```

```python
df.describe().T
```

```python
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Categorical Columns:", cat_cols)
print("Numerical Columns:", num_cols)

for col in cat_cols:
  print("\nValues for counts '{col}':")
  print(df[col].value_counts())
```

```python
df.columns = (
    df.columns
      .str.strip()                       # remove leading/trailing spaces
      .str.lower()                       # optional: lowercase
      .str.replace(r"[()\^]", "", regex=True)  # remove (), ^
      .str.replace(r"\s+", "_", regex=True)    # replace spaces with underscore
)
```

```python
df.isnull().sum()
```

```python
df['peak'] = df['peak'].fillna('NA')
df['peak'].isnull().sum()
print(df.isnull().sum())
```

```python
df['all_time_peak'] = df['all_time_peak'].fillna('NA')
df['all_time_peak'].isnull().sum()
print(df.isnull().sum())
```

```python
df.to_csv('cleaned_my_file.csv', index=False)
```
