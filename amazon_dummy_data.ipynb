## Load Data

### Subtask:
Load the 'amazon_dummy_data_with_errors.csv' file into a pandas DataFrame.


**Reasoning**:
Load the 'amazon_dummy_data_with_errors.csv' file into a pandas DataFrame and display its first few rows to inspect the data.



```python
import pandas as pd

df = pd.read_csv('/content/amazon_dummy_data_with_errors.csv')
df.head()
```

## Data Cleaning

### Subtask:
Perform data cleaning steps, including handling missing values, inconsistent data formats, and duplicates. This step will also include error correction based on the identified issues in the dataset.


**Reasoning**:
To begin the data cleaning process, I will first get an initial overview of the dataset. This includes checking data types and non-null counts, generating descriptive statistics for all columns, and identifying the number of missing values per column. This will help in understanding the current state of the data and guide subsequent cleaning steps.



```python
print("--- DataFrame Info ---")
df.info()

print("\n--- Descriptive Statistics ---")
print(df.describe(include='all').to_markdown(numalign="left", stralign="left"))

print("\n--- Missing Values Count ---")
print(df.isnull().sum().to_frame(name='missing_count').to_markdown(numalign="left", stralign="left"))
```

**Reasoning**:
First, I will standardize the 'product_id' column by replacing the placeholder 'BAD_ID' with `NaN` to uniformly represent missing or invalid product IDs. Then, I will address the 'price' column by converting it to a numeric type, coercing any non-numeric values to `NaN`, and replacing any negative prices with `NaN` to ensure data validity.



```python
import numpy as np

# 2. Standardize Product ID
df['product_id'] = df['product_id'].replace('BAD_ID', np.nan)

# 3. Clean and Convert Price Column
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df.loc[df['price'] < 0, 'price'] = np.nan

print("Product ID and Price columns cleaned.")
```

**Reasoning**:
Now, I will clean and convert the 'rating' column. I'll convert it to a numeric type, coercing non-numeric values to `NaN`, and then ensure all ratings are within the valid range of 1.0 to 5.0, replacing any out-of-range values with `NaN`.



```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df.loc[(df['rating'] < 1.0) | (df['rating'] > 5.0), 'rating'] = np.nan

print("Rating column cleaned and converted.")
```

**Reasoning**:
Next, I will clean and convert the 'reviews_count' column by replacing 'UNKNOWN' with `NaN` and then converting it to a numeric data type, coercing any non-numeric values to `NaN`.



```python
df['reviews_count'] = df['reviews_count'].replace('UNKNOWN', np.nan)
df['reviews_count'] = pd.to_numeric(df['reviews_count'], errors='coerce')

print("Reviews Count column cleaned and converted.")
```

**Reasoning**:
Next, I will clean and convert the 'stock' column to an integer data type, coercing any non-numeric values to `NaN` as per the instructions.



```python
df['stock'] = pd.to_numeric(df['stock'], errors='coerce')

print("Stock column cleaned and converted.")
```

**Reasoning**:
Next, I will clean and convert the 'discount_percent' column by replacing 'INVALID' with `NaN`, converting it to a numeric data type, and replacing values outside the valid range of 0 to 100 with `NaN`.



```python
df['discount_percent'] = df['discount_percent'].replace('INVALID', np.nan)
df['discount_percent'] = pd.to_numeric(df['discount_percent'], errors='coerce')
df.loc[(df['discount_percent'] < 0) | (df['discount_percent'] > 100), 'discount_percent'] = np.nan

print("Discount Percent column cleaned and converted.")
```

**Reasoning**:
Next, I will convert the 'order_date' and 'delivery_date' columns to datetime objects, coercing any unparseable values to `NaN` to handle malformed date strings.



```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')
df['delivery_date'] = pd.to_datetime(df['delivery_date'], errors='coerce')

print("Date columns converted.")
```

**Reasoning**:
To ensure data integrity for critical product identifiers, I will now drop rows where either 'product_name' or 'category' are missing, as these are essential for product identification.



```python
df.dropna(subset=['product_name', 'category'], inplace=True)

print("Rows with missing product_name or category dropped.")
```

**Reasoning**:
Next, I will impute the remaining missing numerical values in 'price', 'rating', 'reviews_count', 'stock', and 'discount_percent'. I will use the median for 'price', 'rating', and 'discount_percent', and 0 for 'reviews_count' and 'stock' as specified in the instructions to maintain data integrity.



```python
df['price'] = df['price'].fillna(df['price'].median())
df['rating'] = df['rating'].fillna(df['rating'].median())
df['reviews_count'] = df['reviews_count'].fillna(0)
df['stock'] = df['stock'].fillna(0)
df['discount_percent'] = df['discount_percent'].fillna(df['discount_percent'].median())

print("Missing numerical values imputed.")
```

**Reasoning**:
Next, I will impute the remaining missing categorical values in 'seller', 'shipment_type', 'brand', 'color', and 'size' columns. I will use the mode of each respective column for imputation to ensure data integrity and consistency.



```python
df['seller'] = df['seller'].fillna(df['seller'].mode()[0])
df['shipment_type'] = df['shipment_type'].fillna(df['shipment_type'].mode()[0])
df['brand'] = df['brand'].fillna(df['brand'].mode()[0])
df['color'] = df['color'].fillna(df['color'].mode()[0])
df['size'] = df['size'].fillna(df['size'].mode()[0])

print("Missing categorical values imputed.")
```

**Reasoning**:
Finally, I will remove any duplicate rows across all columns from the DataFrame to ensure the uniqueness of each entry, which is the last step in the data cleaning process.



```python
df.drop_duplicates(inplace=True)

print("Duplicate rows removed.")
```
